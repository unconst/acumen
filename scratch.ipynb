```python
import random
import aiohttp
import asyncio
import urllib.parse
from async_lru import alru_cache
from typing import Any, Dict, List, Optional

MAX_ROWS_PER_REQUEST = 10

@alru_cache(maxsize=10000)
async def get_split_row_count(dataset: str, config: str = "default", split: str = "train") -> int:
    encoded = urllib.parse.quote(dataset)
    info_url = f"https://datasets-server.huggingface.co/info?dataset={encoded}"
    async with aiohttp.ClientSession() as s:
        async with s.get(info_url) as r:
            r.raise_for_status()
            data = await r.json()
    try:
        return data["dataset_info"][config]["splits"][split]["num_examples"]
    except KeyError:
        raise ValueError(f"Split '{split}' not found in dataset '{dataset}'")

async def fetch_batch(
    session: aiohttp.ClientSession,
    dataset: str,
    config: str,
    split: str,
    offset: int,
    length: int
) -> List[Dict[str, Any]]:
    url = "https://datasets-server.huggingface.co/rows"
    params = {
        "dataset": dataset,
        "config": config,
        "split": split,
        "offset": offset,
        "length": length,
    }
    async with session.get(url, params=params) as r:
        r.raise_for_status()
        data = await r.json()
    return data["rows"]

class AsyncDataLoader:
    """
    Asynchronous data‐loader that samples from a HF dataset server in parallel
    and exposes samples via `async for sample in loader:`.  You can `await loader.reset(seed=...)`
    to reshuffle and start a new pass.
    """

    def __init__(
        self,
        dataset: str,
        config: str = "default",
        split: str = "train",
        sample_size: int = 1,
        seed: Optional[Any] = None,
        max_rows_per_request: int = MAX_ROWS_PER_REQUEST,
        concurrency: Optional[int] = None,
    ) -> None:
        self.dataset = dataset
        self.config = config
        self.split = split
        self.sample_size = sample_size
        self.seed = seed
        self.max_rows = max_rows_per_request
        # number of batches we’ll schedule
        self.num_batches = (sample_size + self.max_rows - 1) // self.max_rows
        # how many fetches to run at once
        self.concurrency = concurrency or self.num_batches

        self._offsets: List[int] = []
        self._lengths: List[int] = []
        self._fetch_tasks: List[asyncio.Task] = []
        self._close_task: Optional[asyncio.Task] = None

        # queue holds individual rows; size bounded by total rows in flight
        self.queue: asyncio.Queue = asyncio.Queue(maxsize=self.num_batches * self.max_rows)
        self._session = aiohttp.ClientSession()
        self._startup: Optional[asyncio.Task] = asyncio.get_event_loop().create_task(
            self.reset(seed)
        )

    async def __anext__(self):
        # on first __anext__, wait for reset to finish
        if self._startup:
            await self._startup
            self._startup = None

        item = await self.queue.get()
        if item is None:
            raise StopAsyncIteration
        return item

    @classmethod
    async def create(
        cls,
        dataset: str,
        config: str = "default",
        split: str = "train",
        sample_size: int = 1,
        seed: Optional[Any] = None,
        max_rows_per_request: int = MAX_ROWS_PER_REQUEST,
        concurrency: Optional[int] = None,
    ) -> "AsyncDataLoader":
        loader = cls(dataset, config, split, sample_size, seed, max_rows_per_request, concurrency)
        await loader.reset(seed)
        return loader

    async def reset(self, seed: Optional[Any] = None) -> None:
        """
        Cancel any in‐flight prefetches, clear the queue, reseed RNG, and
        start fetching fresh batches in the background.
        """
        # cancel old tasks
        for t in self._fetch_tasks:
            t.cancel()
        if self._close_task:
            self._close_task.cancel()

        # fresh queue
        self.queue = asyncio.Queue(maxsize=self.num_batches * self.max_rows)

        # new seed and batch calculation
        self.seed = seed
        total = await get_split_row_count(self.dataset, self.config, self.split)
        n = min(self.sample_size, total - 1)
        random.seed(self.seed)

        self._offsets = sorted(
            random.sample(range(0, total - self.max_rows), self.num_batches)
        )
        self._lengths = [
            min(self.max_rows, n - i * self.max_rows) for i in range(self.num_batches)
        ]

        sem = asyncio.Semaphore(self.concurrency)
        loop = asyncio.get_event_loop()

        # spawn fetch tasks
        self._fetch_tasks = [
            loop.create_task(self._fetch_and_enqueue(offset, length, sem))
            for offset, length in zip(self._offsets, self._lengths)
        ]

        # once all fetches finish, push a sentinel to end iteration
        self._close_task = loop.create_task(self._finalize_queue())

    async def _fetch_and_enqueue(self, offset: int, length: int, sem: asyncio.Semaphore):
        async with sem:
            rows = await fetch_batch(self._session, self.dataset, self.config, self.split, offset, length)
            for row in rows:
                await self.queue.put(row)

    async def _finalize_queue(self):
        await asyncio.gather(*self._fetch_tasks, return_exceptions=True)
        # sentinel None means “we’re done”
        await self.queue.put(None)

    def __aiter__(self):
        return self

    async def __anext__(self) -> Dict[str, Any]:
        item = await self.queue.get()
        if item is None:
            raise StopAsyncIteration
        return item

    async def close(self) -> None:
        """
        Call when you’re done to clean up the underlying HTTP session.
        """
        for t in self._fetch_tasks:
            t.cancel()
        if self._close_task:
            self._close_task.cancel()
        await self._session.close()
```

```python
import random
import aiohttp
import asyncio
import urllib.parse
from async_lru import alru_cache
from typing import Any, Dict, List, Optional

MAX_ROWS_PER_REQUEST = 10

@alru_cache(maxsize=10_000)
async def get_split_row_count(
    dataset: str,
    config: str = "default",
    split: str = "train"
) -> int:
    encoded = urllib.parse.quote(dataset)
    url = f"https://datasets-server.huggingface.co/info?dataset={encoded}"
    async with aiohttp.ClientSession() as sess:
        async with sess.get(url) as resp:
            resp.raise_for_status()
            info = await resp.json()
    try:
        return info["dataset_info"][config]["splits"][split]["num_examples"]
    except KeyError:
        raise ValueError(f"No split {split} in {dataset}/{config}")

async def fetch_batch(
    session: aiohttp.ClientSession,
    dataset: str,
    config: str,
    split: str,
    offset: int,
    length: int,
) -> List[Dict[str, Any]]:
    url = "https://datasets-server.huggingface.co/rows"
    params = {
        "dataset": dataset,
        "config": config,
        "split": split,
        "offset": offset,
        "length": length,
    }
    async with session.get(url, params=params) as resp:
        resp.raise_for_status()
        payload = await resp.json()
        return payload["rows"]

class InfiniteAsyncDataLoader:
    """
    Infinite async loader: keeps buffer topped up with random samples forever.
    
    Usage:
        loader = InfiniteAsyncDataLoader(
            "AI-MO/NuminaMath-TIR",
            sample_batch_size=5,
            buffer_size=20
        )
        # in an async context:
        while True:
            sample = await loader.next()
            print(sample)
    """

    def __init__(
        self,
        dataset: str,
        config: str = "default",
        split: str = "train",
        sample_batch_size: int = 1,
        buffer_size: int = 10,
    ):
        self.dataset = dataset
        self.config = config
        self.split = split
        self.batch_size = sample_batch_size
        self.queue: asyncio.Queue = asyncio.Queue(maxsize=buffer_size)
        self._session = aiohttp.ClientSession()
        self._producer_task = asyncio.get_event_loop().create_task(self._producer())

    async def _producer(self):
        """
        Forever:
          - pick a random offset
          - fetch a batch
          - put each row into the queue (awaiting if full)
        """
        total = await get_split_row_count(self.dataset, self.config, self.split)
        while True:
            offset = random.randrange(0, max(1, total - MAX_ROWS_PER_REQUEST))
            length = MAX_ROWS_PER_REQUEST
            try:
                rows = await fetch_batch(
                    self._session,
                    self.dataset,
                    self.config,
                    self.split,
                    offset,
                    length
                )
                for row in rows:
                    await self.queue.put(row)   # will block if buffer full
            except Exception as e:
                # you can log/handle backoff here
                print(f"Error occurred: {e}")
                await asyncio.sleep(1)

    async def __anext__(self) -> Dict[str, Any]:
        """
        Always returns the next available sample.
        Never raises StopAsyncIteration.
        """
        return await self.queue.get()

    def __aiter__(self):
        return self

    async def next(self) -> Dict[str, Any]:
        """
        Helper so you can write: sample = await loader.next()
        """
        return await self.__anext__()

    async def close(self):
        """
        Clean up the HTTP session and cancel producer.
        """
        self._producer_task.cancel()
        await self._session.close()
```
